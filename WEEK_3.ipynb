Matrix Operation-R


```r
x <- matrix(
    nrow=4,
    ncol=2,
    data=c(1,2,3,4,5,6,7,8)
)
print(x)
```

```r
x <- matrix(
  nrow = 4,
  ncol = 2,
  data = c(1,2,3,4,5,6,7,8),
  byrow = FALSE
)

print(x)
```

```r
y <- matrix(
  nrow = 4,
  ncol = 2,
  data = c(1,2,3,4,5,6,7,8),
  byrow = TRUE
)

print(y)
```

```r
print(x[3,2])
```

```r
print(x[1,1])
print(x[2,1])
print(x[4,2])
```

```r
print(dim(x))
```

Number of rows

```r
print(nrow(x))
```

Number of columns

```r
print(ncol(x))
```

 Find mode/type

```r
print(mode(x))
```

```r
print(attributes(x))
```

Check whether x is a matrix

```r
print(is.matrix(x))
```

Convert vector to matrix

```r
v <- c(1,2,3,4)
v_matrix <- as.matrix(v)
print(v_matrix)
```
