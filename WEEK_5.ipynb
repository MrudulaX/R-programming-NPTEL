# Lecture 18 — Missing Data Handling

The notes introduce `NA` as R's representation of a missing observation, `is.na()` for detecting missing values, `which(is.na())` for locating them, `sum(is.na())` for counting them, `complete.cases()` for identifying complete observations, and `na.omit()` for removing missing observations. fileciteturn4file0L13-L23

## Missing values: R `NA` vs Python `None` / `NaN`

For numerical work, NumPy uses `np.nan` as a common representation of a missing numeric value.

```python
x = np.array([11.0, np.nan, 13.0, np.nan])
print(x)
print("Missing-value mask:", np.isnan(x))
```

## Detect missing values — equivalent of `is.na()`

```python
x = np.array([11.0, np.nan, 13.0, np.nan])
is_missing = np.isnan(x)

print("x:", x)
print("is.na(x) equivalent:", is_missing)
```

## Mean with missing values

The R notes show that `mean(x)` returns `NA` when the vector contains missing values, while `mean(x, na.rm = TRUE)` removes missing values before calculating the mean. fileciteturn4file0L39-L44

```python
x = np.array([11.0, np.nan, 13.0, np.nan])

print("Regular NumPy mean:", np.mean(x))
print("Mean ignoring NaN (R na.rm=TRUE equivalent):", np.nanmean(x))
```

## NA versus NULL

The notes distinguish `NA` from `NULL`: `NA` is a placeholder for something that exists but is missing, whereas `NULL` represents something that never existed. fileciteturn4file0L55-L60

In Python, `None` is often used for the absence of an object/value. It is not identical to R's `NA`.

```python
missing_value = None
numeric_missing = np.nan

print("None:", missing_value)
print("NaN:", numeric_missing)
```

## Locate missing values — equivalent of `which(is.na(x))`

R's `which(is.na(x))` returns the positions containing `NA`. The notes demonstrate positions 2 and 4 for `c(11, NA, 13, NA)`. fileciteturn4file0L64-L71

Python uses zero-based indexing, so the corresponding zero-based positions are 1 and 3.

```python
x = np.array([11.0, np.nan, 13.0, np.nan])

zero_based_positions = np.where(np.isnan(x))[0]
one_based_positions = zero_based_positions + 1

print("Python positions (0-based):", zero_based_positions)
print("R-style positions (1-based):", one_based_positions)
```

## Count missing values — equivalent of `sum(is.na(x))`

```python
x = np.array([11.0, np.nan, 13.0, np.nan])

print("Number of missing values:", np.isnan(x).sum())
```

## Complete cases — equivalent of `complete.cases()`

For a one-dimensional numeric array, a complete case is a value that is not missing.

```python
x = np.array([11.0, np.nan, 13.0, np.nan])

complete = ~np.isnan(x)
print("Complete-case mask:", complete)
```

## Remove missing values — equivalent of `na.omit()`

For numeric NumPy arrays, `x[~np.isnan(x)]` keeps only non-missing values.

```python
x = np.array([11.0, np.nan, 13.0, np.nan])
y = x[~np.isnan(x)]

print("Original:", x)
print("After removing missing values:", y)
print("Mean after removal:", np.mean(y))
```

## Missing values in a pandas DataFrame

For tabular data, pandas provides `isna()`, `notna()`, `dropna()` and related operations.

```python
df = pd.DataFrame({
    "A": [11, np.nan, 13, np.nan],
    "B": [5, 6, np.nan, 8]
})

print("DataFrame:")
display(df)

print("Missing-value mask:")
display(df.isna())

print("Missing values per column:")
print(df.isna().sum())

print("Complete rows:")
display(df.dropna())
```

# Lecture 19 — Conditional Execution: `if` and `if-else`

The notes introduce control structures and explain that `if` executes its block when a logical condition is `TRUE`; otherwise the block is ignored. fileciteturn4file0L145-L160

## Python `if` — basic example

R:
`if (x > 4) x * 3`

Python uses the same basic conditional idea with indentation.

```python
x = 5

if x > 4:
    print(x * 3)
```

## `if` with a false condition

The R example with `x = 3` produces no output because the condition `x > 4` is false. fileciteturn4file0L177-L185

```python
x = 3

if x > 4:
    print(x * 3)

print("The if block was skipped because x > 4 is False.")
```

## Printing based on a condition

```python
x = 6

if x > 3:
    print("The value is more than 3")
```

## `if-else`

R's `if (...) { ... } else { ... }` maps directly to Python's `if ... else ...`. The notes use examples where `x == 3` causes `x = x - 1`, otherwise `x = 2*x`. fileciteturn4file0L223-L256

```python
x = 5

if x == 3:
    x = x - 1
else:
    x = 2 * x

print(x)
```

### Same example with `x = 3`

```python
x = 3

if x == 3:
    x = x - 1
else:
    x = 2 * x

print(x)
```

## `if-else` for a message

```python
x = 2

if x > 3:
    print("The value is more than 3")
else:
    print("The value is less than 3")
```

## Logical operators

The notes mention `&&` and `||` for compound scalar conditions in R. Python uses:
- `and` for scalar logical AND
- `or` for scalar logical OR
- `not` for logical negation

```python
x = 6

if x > 3 and x < 10:
    print("x is between 3 and 10")
```

# Lecture 20 — Nested `if-else-if` and `ifelse`

The notes explain that `if ... else if ... else` handles more than two alternatives. fileciteturn4file0L346-L364

## Nested / chained conditional

```python
x = 5

if x == 3:
    x = x - 1
elif x < 3:
    x = x + 5
else:
    x = 2 * x

print(x)
```

### Example with `x = 2`

```python
x = 2

if x == 3:
    x = x - 1
elif x < 3:
    x = x + 5
else:
    x = 2 * x

print(x)
```

### Example with `x = 3`

```python
x = 3

if x == 3:
    x = x - 1
elif x < 3:
    x = x + 5
else:
    x = 2 * x

print(x)
```

## Vectorized `ifelse` — NumPy `where`

R's `ifelse(test, yes, no)` performs element-wise conditional evaluation. The notes show `ifelse(x < 6, x^2, x+1)` for `x = 1:10`. fileciteturn4file0L423-L449

```python
x = np.arange(1, 11)

result = np.where(x < 6, x**2, x + 1)

print("x:", x)
print("Result:", result)
```

## Vectorized even/odd classification

The notes use `ifelse(x %% 2 == 0, "even number", "odd number")`. In Python/NumPy, `%` is the modulo operator and `np.where` provides the vectorized conditional equivalent. fileciteturn4file0L456-L468

```python
x = np.array([7, 9, 8, 4])

result = np.where(x % 2 == 0, "even number", "odd number")

print(result)
```

## Python list-comprehension alternative for `ifelse`

```python
x = [7, 9, 8, 4]

result = ["even number" if value % 2 == 0 else "odd number" for value in x]

print(result)
```

# Lecture 21 — `switch` and `which`

The notes describe `switch()` as a multiway branch statement that tests a variable against several cases. They also introduce `which()` as a function that returns positions where a logical vector is `TRUE`. fileciteturn4file0L496-L515

## R `switch()` with integer cases

The notes show:
- `switch(2, "apple", "banana", "orange")` → `"banana"`
- `switch(1, "apple", "banana", "orange")` → `"apple"` fileciteturn4file0L551-L557

Python has no exact built-in equivalent, so a dictionary lookup is a clear translation.

```python
cases = {
    1: "apple",
    2: "banana",
    3: "orange"
}

expr = 2
print(cases.get(expr))
```

## String-based `switch()` equivalent

The R notes also show named cases such as `"colour" = "blue"` and `"volume" = 50`. fileciteturn4file0L561-L572

```python
cases = {
    "colour": "blue",
    "gender": "male",
    "volume": 50
}

expr = "volume"
print(cases.get(expr))
```

## Handling a default case

R's `switch()` can return an unnamed/default value when there is no match. In Python, `dict.get(key, default)` provides a similar pattern.

```python
cases = {
    "colour": "blue",
    "gender": "male",
    "volume": 50
}

expr = "size"
print(cases.get(expr, "No match"))
```

## `which()` — find positions where a condition is true

The notes use `x = c(10,15,8,14,6,12)` and demonstrate `which(x == 14)`, `which(x != 12)` and `which(x > 10)`. fileciteturn4file0L588-L612

```python
x = np.array([10, 15, 8, 14, 6, 12])

def which(condition):
    return np.where(condition)[0] + 1  # R-style 1-based positions

print("x:", x)
print("which(x == 14):", which(x == 14))
print("which(x != 12):", which(x != 12))
print("which(x > 10):", which(x > 10))
```

## `which.min()` and `which.max()`

The notes demonstrate `which.min(x)` and `which.max(x)` on a 3×3 matrix. Python's `np.argmin()` and `np.argmax()` return zero-based flat indices, so `+1` gives an R-style flat position. fileciteturn4file0L616-L627

```python
x = np.arange(1, 10).reshape(3, 3, order="F")

print("Matrix:")
print(x)

min_position_r = np.argmin(x) + 1
max_position_r = np.argmax(x) + 1

print("R-style which.min position:", min_position_r)
print("R-style which.max position:", max_position_r)
```

## `which()` on a matrix

The R notes use `which(x %% 2 == 1, arr.ind = TRUE)` to obtain row/column positions of odd values. fileciteturn4file0L632-L650

```python
x = np.arange(1, 10).reshape(3, 3, order="F")

print("Matrix:")
print(x)

odd_positions = np.argwhere(x % 2 == 1)

print("Python row/column positions (0-based):")
print(odd_positions)

print("R-style row/column positions (1-based):")
print(odd_positions + 1)
```

# R → Python Quick Reference

|---|---|
| `NA` | `np.nan` for numeric missing values |
| `is.na(x)` | `np.isnan(x)` / `pd.isna(x)` |
| `which(is.na(x))` | `np.where(np.isnan(x))[0] + 1` for R-style positions |
| `sum(is.na(x))` | `np.isnan(x).sum()` |
| `complete.cases(x)` | `~np.isnan(x)` for numeric vectors |
| `na.omit(x)` | `x[~np.isnan(x)]` |
| `if (...)` | `if ...:` |
| `if (...) else` | `if ... else ...` |
| `else if` | `elif` |
| `ifelse()` | `np.where()` |
| `switch()` | dictionary lookup / `match-case` |
| `which()` | `np.where()` |
| `which.min()` | `np.argmin()` |
| `which.max()` | `np.argmax()` |
| R indexing | 1-based |
| Python indexing | 0-based |
