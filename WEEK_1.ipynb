### 4. Demonstration of graphics

The R notes use `demo(persp)` for 3D surface plots and `demo(graphics)` for graphics
demonstrations. Python can demonstrate plotting with Matplotlib.


```python
# 2D graphics demonstration
import matplotlib.pyplot as plt

x = list(range(0, 11))
y = [value ** 2 for value in x]

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("x²")
plt.title("Python graphics demonstration")
plt.show()
```

```python
# Simple 3D surface demonstration corresponding conceptually to R's persp()
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 40)
y = np.linspace(-3, 3, 40)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("3D surface plot")
plt.show()
```

The notes explain that R packages are collections of programs and datasets that extend

Python has a similar ecosystem of modules and packages.


### Importing a Python package

```python
# Standard-library package/module
import math

print(math.sqrt(25))
print(math.pi)
```

### Package description / metadata

```python
# Python equivalent of inspecting package/module information
import sys

print("Python version:", sys.version)
print("Python executable:", sys.executable)
```

### Listing available names in a package

```python
import math

print(dir(math))
```

### Installed Python packages

```python
# Show installed distributions in the current Python environment.
import importlib.metadata as metadata

packages = sorted(
    [(dist.metadata.get("Name", ""), dist.version) for dist in metadata.distributions()],
    key=lambda item: item[0].lower()
)

print("Number of installed distributions:", len(packages))
print("First 30:")
for name, version in packages[:30]:
    print(f"{name} - {version}")
```

# Lecture 4 — Command Line and Data Editor

The notes explain that R commands can be entered directly at the command line and that
multi-line programs are more conveniently stored in scripts. R's built-in editor and
RStudio are discussed.

Python has the same general distinction between interactive execution and script/notebook
execution.


## Interactive command-line style

```python
# These are individual Python statements, similar to entering commands interactively.
x = 20
y = 2 * x
z = x + y

print(x)
print(y)
print(z)
```

## Multi-line program / script style

```python
def calculate_values(x):
    y = 2 * x
    z = x + y
    return y, z

result = calculate_values(20)
print("y, z =", result)
```

## Working with a Python script

A `.py` file can be created and executed separately from a notebook. The following
cell creates a small example script so you can see the relationship between a notebook
and a normal source file.


```python
script_code = '''def square(x):
    return x * x

for value in [1, 2, 3, 4, 5]:
    print(value, square(value))
'''

with open("example_script.py", "w", encoding="utf-8") as f:
    f.write(script_code)

print("Created example_script.py")
```

```python
# Execute the example script from Python.
import subprocess
import sys

result = subprocess.run(
    [sys.executable, "example_script.py"],
    capture_output=True,
    text=True
)

print(result.stdout)
```
