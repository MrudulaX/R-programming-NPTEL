## Decimal sequences

The R `:` operator can also create sequences with a unit increment/decrement when the endpoints are numeric. fileciteturn3file0L54-L62


```python
def r_colon_decimal(start, end, step=1.0):
    values = []
    x = float(start)
    if step > 0:
        while x <= end + 1e-12:
            values.append(round(x, 10))
            x += step
    else:
        while x >= end - 1e-12:
            values.append(round(x, 10))
            x += step
    return values

print(r_colon_decimal(1.23, 10))
print(r_colon_decimal(1.23, 10.54))
print(r_colon_decimal(10.54, 2.23, -1))
print(r_colon_decimal(-1.23, -10, -1))
print(r_colon_decimal(-5.23, 6, 1))
```

## `seq()` equivalents

The notes show `seq(10)` producing 1 through 10 and examples where `from`, `to`, and `by` are specified. fileciteturn3file0L82-L98


```python
def seq_r(from_value=1, to_value=1, by=None, length_out=None):
    if length_out is not None:
        if length_out == 1:
            return [from_value]
        return np.linspace(from_value, to_value, length_out).tolist()
    if by is None:
        by = 1 if to_value >= from_value else -1
    values = []
    x = from_value
    if by > 0:
        while x <= to_value + 1e-12:
            values.append(x)
            x += by
    else:
        while x >= to_value - 1e-12:
            values.append(x)
            x += by
    return values

print("seq(10):", seq_r(1, 10))
print("seq(1, x, x/10), x=2:", seq_r(1, 2, 0.2))
print("seq(0, x, x/10), x=50:", seq_r(0, 50, 5))
```

## Date sequences by year

R's date `seq()` supports increments such as `"day"`, `"week"`, `"month"`, `"quarter"` and `"year"`. fileciteturn3file0L173-L184


```python
def date_sequence(start, count, unit):
    start = datetime.strptime(start, "%Y-%m-%d").date()
    result = []
    current = start

    for _ in range(count):
        result.append(current)
        if unit == "day":
            current += timedelta(days=1)
        elif unit == "week":
            current += timedelta(weeks=1)
        elif unit == "month":
            current += relativedelta(months=1)
        elif unit == "year":
            current += relativedelta(years=1)
        else:
            raise ValueError("Use day, week, month, or year")
    return result

print("Year sequence:")
print(date_sequence("2010-01-01", 8, "year"))

print("\nDay sequence:")
print(date_sequence("2017-01-01", 6, "day"))

print("\nMonth sequence:")
print(date_sequence("2017-01-01", 6, "month"))

print("\nYear sequence:")
print(date_sequence("2017-01-01", 6, "year"))
```

## Reverse monthly sequence

The notes generate a monthly sequence from `2017-01-01` backward to `2016-01-01`. fileciteturn3file0L227-L236


```python
start = date(2017, 1, 1)
end = date(2016, 1, 1)

dates_backward = []
current = start

while current >= end:
    dates_backward.append(current)
    current -= relativedelta(months=1)

print(dates_backward)
```

# Lecture 28 – Repeats

R's `rep()` replicates numeric values, text, or vector values. Important forms include:
- `rep(x, times=n)` – repeat the whole vector
- `rep(x, each=n)` – repeat each element
- `rep(x, length.out=n)` – generate exactly the requested output length. fileciteturn3file0L300-L316


```python
x = [1, 2, 3, 4]

print("rep(3.5, times=10):", [3.5] * 10)
print("rep(1:4, 2):", x * 2)

print("rep(x, times=3):", x * 3)
print("rep(x, each=3):", [value for value in x for _ in range(3)])
```

## `each` and `times` together

The notes demonstrate `rep(1:4, each=2, times=3)` and show the effect of repeating each element before repeating the whole pattern. fileciteturn3file0L351-L360


```python
pattern = [value for value in range(1, 5) for _ in range(2)]
result = pattern * 3

print("Pattern:", pattern)
print("Repeated 3 times:", result)
```

## Different repetition count for each element

The notes show `rep(1:4, 2:5)`, where the first element is repeated 2 times, the second 3 times, etc. fileciteturn3file0L368-L373


```python
values = [1, 2, 3, 4]
counts = [2, 3, 4, 5]

result = []
for value, count in zip(values, counts):
    result.extend([value] * count)

print(result)
```

## Repetition counts generated by a sequence

The notes use `ans = seq(from=2, to=8, by=2)` and then `rep(1:4, ans)`. fileciteturn3file0L375-L382

# Lecture 29 – Sorting, Ordering and Mode

R `sort()` sorts a vector in ascending order by default, with `decreasing=TRUE` for descending order. fileciteturn3file0L442-L453


```python
y = [8, 5, 7, 6]

print("Original:", y)
print("Ascending:", sorted(y))
print("Descending:", sorted(y, reverse=True))
```

## `order()` equivalent

R's `order()` returns the positions/indices that would sort a vector. For `y = c(9,8,5,7,6)`, the notes show `order(y) = 3,5,4,2,1`. fileciteturn3file0L482-L506


```python
y = [9, 8, 5, 7, 6]

order_zero_based = sorted(range(len(y)), key=lambda i: y[i])
order_r_style = [i + 1 for i in order_zero_based]

print("y:", y)
print("Python zero-based order:", order_zero_based)
print("R-style 1-based order:", order_r_style)

order_desc = sorted(range(len(y)), key=lambda i: y[i], reverse=True)
order_desc_r_style = [i + 1 for i in order_desc]

print("R-style decreasing order:", order_desc_r_style)
print("Values in sorted order:", [y[i] for i in order_zero_based])
```

## R `mode()` versus statistical mode

In R, `mode()` describes how an object is stored in memory (for example numeric, character, list, function), rather than calculating the most frequent value. The notes give examples for numbers, character strings, factors, lists, data frames and functions. fileciteturn3file0L514-L540

Python does not have an exact `mode()` equivalent with the same R meaning. `type()` is the closest general inspection tool.


```python
objects = {
    "number": 2.432,
    "numeric_vector": [3, 4, 5, 6, 7, 8],
    "character": "India",
    "character_vector": ["India", "CANADA"],
    "list": ["India", "USA"],
    "function": print
}

for name, obj in objects.items():
    print(f"{name:18} -> {type(obj)}")
```

### Python statistical mode (different concept)

If you want the **most frequent value** of a dataset, Python's `Counter` can be used. This is a statistical mode, not R's `mode()` storage-class concept.


```python
data = [1, 2, 2, 3, 2, 4, 3, 2]
frequency = Counter(data)

print("Frequencies:", frequency)
print("Statistical mode:", frequency.most_common(1)[0][0])
```

# R → Python Quick Reference

|---|---|
| `1:10` | `range(1, 11)` |
| `seq()` | `range()`, `numpy.arange()`, `numpy.linspace()` |
| `Sys.time()` | `datetime.now()` |
| `Sys.Date()` | `date.today()` |
| Date `seq(..., by="days")` | `timedelta` loop |
| Date `seq(..., by="months")` | `relativedelta(months=1)` |
| `letters` | `string.ascii_lowercase` |
| `LETTERS` | `string.ascii_uppercase` |
| `rep(x, times=n)` | `x * n` |
| `rep(x, each=n)` | list comprehension |
| `rep(..., length.out=n)` | repeat then slice / `itertools.cycle` |
| `sort(x)` | `sorted(x)` |
| `sort(x, decreasing=TRUE)` | `sorted(x, reverse=True)` |
| `order(x)` | sorted index positions |
| R `mode()` | `type()` for storage/type inspection |
| R statistical mode | `Counter(...).most_common()` |
| `list()` | Python `list` |
| `x[[i]]` | `x[i-1]` conceptually, with Python's 0-based indexing |
