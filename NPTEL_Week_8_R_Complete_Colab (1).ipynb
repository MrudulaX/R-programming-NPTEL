# 2. Vector Indexing

The notes create `x = 1:10` and select values using logical conditions.


```r
x <- 1:10

print(x)
print(x[(x > 5)])

# %% means modulo/remainder
print(x[(x %% 2 == 0)])  # even values
print(x[(x %% 2 == 1)])  # odd values
```

```r
x <- 1:10
x[5] <- NA

print(x)

y <- x[!is.na(x)]
print(y)

print(mean(x))
print(mean(y))
```

# 3. Factors

## 3.1 Categorical variables

The notes distinguish quantitative variables from qualitative/categorical variables.

Examples:
- Quantitative: height in meters, such as 1.65, 1.76, …
- Qualitative: gender (Male/Female), performance (Excellent/Good/Average/Bad)

In R, categorical variables are represented using **factors**.


## 3.2 Factor concept with ball colours

The notes use:
- Red = 1
- Blue = 2
- Green = 3

Observed colours:
`Red, Green, Green, Blue, Red`

A factor stores categories as levels with internal numeric codes.


```r
colours <- c("Red", "Green", "Green", "Blue", "Red")

colour_factor <- factor(
  colours,
  levels = c("Red", "Blue", "Green")
)

print(colour_factor)
print(levels(colour_factor))
print(unclass(colour_factor))
```

## 3.3 Defining a factor

The basic form is:

`factor(x)`

If the possible levels need to be specified:

`factor(x, levels)`

The notes also describe the optional `labels` and `exclude` arguments.


```r
# Basic factor
x <- c("A", "B", "A", "C", "B")
f1 <- factor(x)

print(f1)
print(levels(f1))

# Explicit levels
f2 <- factor(x, levels = c("A", "B", "C"))

print(f2)
print(levels(f2))
```

## 3.4 Die-roll example

The notes roll a die seven times:

`1, 4, 3, 5, 4, 2, 4`

Possible die faces are 1 to 6, and the labels are:
`one, two, three, four, five, six`.


```r
y <- c(1, 4, 3, 5, 4, 2, 4)

possible.dieface <- c(1, 2, 3, 4, 5, 6)

labels.dieface <- c(
  "one", "two", "three",
  "four", "five", "six"
)

facy <- factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

print(facy)
print(levels(facy))
```

## 3.5 Character vector versus factor

Observe that `y` contains the original numeric observations while `facy` is a factor with labelled levels.


```r
y <- c(1, 4, 3, 5, 4, 2, 4)

print(y)
print(facy)

print(class(y))
print(class(facy))
```

# 4. Factors — `class()` and `unclass()`

## 4.1 Converting a vector to a factor with `as.factor()`

The notes use a numeric vector containing values 1 through 6 and convert it using `as.factor()`.


```r
x <- c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)

print(x)

y <- as.factor(x)

print(y)
print(levels(y))
```

## 4.2 Character values become alphabetically ordered levels

The notes use `lemonade`, `juice`, and `water`.


```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water")
)

print(x)
print(levels(x))
```

## 4.3 `class()` function

The notes explain that `class()` reports the class of an R object.

Examples include:
- `class(9)` → numeric
- `class("9")` → character
- `class(print)` → function
- a matrix has class `matrix` and `array`


```r
print(class(9))
print(class("9"))
print(class(print))

x <- matrix(nrow = 2, ncol = 2, data = 1:4)
print(x)
print(class(x))
```

## 4.4 `unclass()` function

`unclass()` temporarily removes the effects of an object's class.

The notes demonstrate this with a factor.


```r
brands <- c("A", "A", "B", "B", "B", "B", "C")

print(brands)

brands_fac <- factor(brands)

print(brands_fac)
print(unclass(brands_fac))
```

## 4.5 Converting factor codes back to category names

The notes create:

`colours = c("blue", "green", "red")`

Then use the integer codes from `unclass(brands_fac)` to index the colour vector.


```r
brands <- c("A", "A", "B", "B", "B", "B", "C")
brands_fac <- factor(brands)

colours <- c("blue", "green", "red")

codes <- unclass(brands_fac)

print(codes)
print(colours[codes])
```

## 4.6 Another `unclass()` example

For the factor created from `lemonade`, `juice`, and `water`, the internal codes depend on the order of the factor levels.


```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water")
)

print(x)
print(unclass(x))
print(levels(x))
```

## 4.7 Controlling the factor level order

The notes explicitly set:

`levels = c("water", "juice", "lemonade")`

This changes the internal numeric codes associated with the categories.


```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water"),
  levels = c("water", "juice", "lemonade")
)

print(x)
print(unclass(x))
print(levels(x))
```

## 4.8 Ordered factors

The notes create an ordered income factor with the order:

`low < medium < high`


```r
income <- ordered(
  c("high", "high", "low", "medium", "medium"),
  levels = c("low", "medium", "high")
)

print(income)
print(levels(income))
print(unclass(income))
```

The notes cover four important commands:

The examples below reproduce the demonstrated `print()` and `format()` operations and also give small demonstrations of `cat()` and `paste()` because they are listed as important display/formatting commands in the notes.


## 5.1 `print()` function

`print()` prints its argument. It is a generic command available for different object classes.


```r
print(sqrt(2))
print(sqrt(2), digits = 5)
print(sqrt(2), digits = 10)
```

## 5.2 Printing character values

The notes demonstrate printing one string, a character vector, and a mixed vector.


```r
print("apple")
print(c("apple", "banana"))
print(c("apple", "banana", 6, 10))
```

## 5.3 `format()` function

`format()` prepares an R object for pretty printing.

Important arguments from the notes:
- `digits`: number of significant digits
- `nsmall`: minimum number of digits after the decimal point
- `justify`: left, right, centre, or none
- `width`: field width


```r
print(format(0.5, digits = 10, nsmall = 15))
```

## 5.4 Formatting with width and centre justification

The notes demonstrate width 7 and width 14.


```r
print(format(
  c("A", "BB", "CCC", "DDDD"),
  width = 7,
  justify = "centre"
))

print(format(
  c("A", "BB", "CCC", "DDDD"),
  width = 14,
  justify = "centre"
))
```

## 5.5 Formatting with different `justify` values

The notes compare:
- centre
- left
- right
- none


```r
values <- c("A", "BB", "CCC", "DDDD")

print(format(values, width = 7, justify = "centre"))
print(format(values, width = 7, justify = "left"))
print(format(values, width = 7, justify = "right"))
print(format(values, width = 7, justify = "none"))
```

## 5.6 Formatting a matrix

```r
x <- matrix(
  nrow = 3,
  ncol = 2,
  data = 1:6,
  byrow = TRUE
)

print(x)
```

## 5.7 Formatting large quantities with `big.mark`

The notes use commas to separate groups of digits.


```r
print(format(1234567, big.mark = ","))
print(format(12345678, big.mark = ","))
print(format(123456789, big.mark = ","))
```

## 5.8 Formatting large quantities with spaces

The notes also demonstrate using a space as the grouping separator.


```r
print(format(123456789, big.mark = " "))
```

## 5.9 `cat()` — display text

`cat()` is one of the important string display commands listed in the notes. It prints text without the usual vector-style `print()` formatting.


```r
cat("Hello", "
")
cat("Apple", "Banana", "
")
cat("Name:", "Mrudula", "Score:", 95, "
")
```

## 5.10 `paste()` — combine strings

`paste()` is also listed among the important string-formatting commands. It combines character values.


```r
first_name <- "Mrudula"
last_name <- "Patil"

print(paste(first_name, last_name))
print(paste("R", "Programming", sep = "-"))
print(paste(c("A", "B", "C"), collapse = ", "))
```
